```python
import torch
```

```python
# Input Data
i = torch.tensor([.05, .1])
```

```python
# Weight and Bias Initialization

# Layer 1
w_1 = torch.tensor([.15, .20, .25, .30], requires_grad = True)
b1 = torch.tensor([.35], requires_grad = True)

# Layer 2
w_2 = torch.tensor([.40, .45, .50, .55], requires_grad = True)
b2 = torch.tensor([.60], requires_grad = True)  
```

```python
# Actual Output
y1 = torch.tensor([.01])
y2 = torch.tensor([.99])
```

```python
# Forward Pass
# Layer 1
z1 = i[0] * w_1[0] + i[1] * w_1[2] + b1
z2 = i[0] * w_1[1] + i[1] * w_1[3] + b1
```

```python
# Forward Pass
# Passing through activation function
a1 = torch.sigmoid(z1)
a2 = torch.sigmoid(z2)
```

```python
# Forward Pass
# Output Layer
z3 = a1 * w_2[0] + a2 * w_2[1] + b2
z4 = a1 * w_2[2] + a2 * w_2[3] + b2
```

```python
# Forward Pass
# Passing through activation function
y1_pred = torch.sigmoid(z3)
y2_pred = torch.sigmoid(z4)
```

```python
# Error Calculation
E01 = ((y1 - y1_pred) ** 2) / 2
E02 = ((y2 - y2_pred) ** 2) / 2
```

```python
# Total Error
E = E01 + E02
```

```python
# Backward Pass
E.backward()
```
